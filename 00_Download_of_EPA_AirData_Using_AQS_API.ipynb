Replace EMAIL and API_KEY with your information. This code is to download a .csv file, but can be modified for .txt.

If utilizing data other than the daily data by state, ensure to change the 'url =' variable.

```python
import requests
import pandas as pd
from pathlib import Path

EMAIL = "[actual email redacted]"
API_KEY = "[actual API key redacted]"

output_dir = Path("/content/drive/MyDrive/epa_airdata") #or custom file location
output_dir.mkdir(parents=True, exist_ok=True)

#change this to your desired file name:
output_csv = output_dir / "IL_daily_PM25_Jun_Sep_2015_2025_AQS_API.csv"

years = range(2015, 2025)
dfs = []

for year in years:
    bdate = f"{year}0601"  # June 1
    edate = f"{year}0930"  # September 30

    print(f"Fetching {year} data...")

    url = "https://aqs.epa.gov/data/api/dailyData/byState"
    params = {
        "email": EMAIL,
        "key": API_KEY,
        "state": "17",      #If selecting another state, change state code (currently for IL)
        "param": "88101",   #If selecting data other than PM2.5, change this (currently PM2.5 data code)
        "bdate": bdate,
        "edate": edate
    }

    response = requests.get(url, params=params)
    response.raise_for_status()
    result = response.json()

    #Double checking that the data downloaded:
    if "Data" in result and result["Data"]:
        dfs.append(pd.DataFrame(result["Data"]))
    else:
        print(f"No data returned for {year}")

if dfs:
    df_all = pd.concat(dfs, ignore_index=True)

    df_all.to_csv(output_csv, index=False)
    print(f"Saved file with {len(df_all):,} rows to:\n{output_csv}")
else:
    print("No data was downloaded")
```
